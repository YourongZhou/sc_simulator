Import modules and simulate data

```python
import numpy as np
import scanpy as sc
import seaborn as sns

import sys
sys.path.append('../simulator')
from hiererchical_simulator import get_data_and_visualize

data_df, adata = get_data_and_visualize(tree_id=100)
sc.pp.normalize_total(adata, target_sum=1e4)
```

Normalize data by cell and tanspose the data such that a row is a gene.

```python
X = adata.X.astype(np.float32)
X_normed = X / X.sum(axis=1)[None,:].T * 1e4
X_normed = X_normed.T
```

Import the CME module

```python
sys.path.append('../CME')
import CME_module
from CME_module import CME, CME_corrected
```

CME CPU code examples

```python
# Compute CME on CPU
CME_module.cme_config["device"] = "CPU"
cme = CME(X_normed)
sns.heatmap(cme, cmap='viridis')
```

```python
# Correct CME on CPU
CME_module.cme_config["device"] = "CPU"
cme_corrected, pval = CME_corrected(X_normed)
sns.heatmap(cme, cmap='viridis')
```

CME GPU code examples

```python
# Compute CME on CPU
CME_module.cme_config["device"] = "GPU"
cme = CME(X_normed, normalize=True)
sns.heatmap(cme, cmap='viridis')
```

```python
# Correct CME on GPU
CME_module.cme_config["device"] = "GPU"
cme_corrected, pval = CME_corrected(X_normed)
sns.heatmap(cme_corrected, cmap='viridis')
```
